Program 1: To implement push, pop, peek, and display operations using a stack.
Process:
1. Create an empty list as a stack.
2. Push elements to the top.
3. Pop the top element if the stack is not empty.
4. Display the top element or all elements when requested.

```python
stack=[]
while True:
    print('1. Push')
    print('2. Pop')
    print('3. peek')
    print('4. Display')
    print('5. Exit')

    choice=int(input("Enter your choice"))

    if choice==1:
        value=int(input("Enter value"))
        stack.append(value)
        print("Value pushed")

    elif choice==2:
        if not stack:
            print("Stack underflow")
        else:
            print("value popped",stack.pop())
    elif choice==3:
        if not stack:
            print("stack empty")
        else:
            print(stack[-1])
    elif choice==4:
        if not stack:
            print("Empty stack")
        else:
            print("Stacks are in order of: ",stack[::-1])

    elif choice==5:
        break

    else:
        print("Wrong choice")
```

Program 2. Check Validity of a Parenthesized Arithmetic Expression
  

```python
def isvalid(expression):
    stack=[]
    for ch in expression:
        if ch=="(":
            stack.append(ch)
        elif ch==")":
            if not stack:
                return False
            else:
                stack.pop()
    return len(stack)==0
expression=input()
if isvalid(expression):
    print("it was valid expression")
else:
    print("Not valid")
```

program 3. Check Validity of a Bracketed Arithmetic Expression
 

```python
def isvalid(expression):
    stack=[]
    for ch in expression:
        if ch=="(" or ch=="{" or ch=="[":
            stack.append(ch)
        elif ch==")":
            if not stack or stack[-1]!="(":
                return False
            stack.pop()
        elif ch=="]":
            if not stack or stack[-1]!="[":
                return False
            stack.pop()
        elif ch=="}":
            if not stack or stack[-1]!="{":
                return False
            stack.pop()
    return len(stack)==0
expression=input()
if isvalid(expression):
    print("it was valid expression")
else:
    print("Not valid")
```

Program 4. Check Whether a Number Is a Palindrome Using a Stack



```python
def isPalindrome(expression):
    stack=[]
    for ch in expression:
        stack.append(ch)

    rslt=""

    i=len(stack)
    while stack:
        rslt+=stack.pop()
    
    if rslt==expression:
        return True
    else:
        return False


print(isPalindrome("123321"))
print(isPalindrome("1221"))
```

Program 5. Reverse a String Using a Stack




```python
def rev(exp):
    stack=[]
    for ch in exp:
        stack.append(ch)
    res=""
    while stack:
        res+=stack.pop()
    return res
rev("ayush")
```
