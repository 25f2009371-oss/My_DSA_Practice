Program 18. Factorial of a Number Using Recursion

Formula: \(n! = n \times (n-1)!\), with \(0! = 1\).

```python
def factorial(n):
    if n<0:
        return "Factorial dont defined"
    elif n==0 or n==1:
        return 1
    return n*factorial(n-1)

print(factorial(5))
print(factorial(-1))
```

Program 19: Calculate \(a^n\) Using Recursion

formula= 2^4 =2*2*2*2

```python
def power(a,n):

    if n==0:
        return 1
    return a*power(a,n-1)
power(5,4)
```

Program 20. Find the nth Fibonacci Number Using Recursion and Improve Runtime


```python
def fib(n):

    a,b=0,1
    for _ in range(n):
        a,b=b,a+b
    return a
fib(7)
```

Program 21. GCD of Two Numbers Using Recursion

```python
def gcd(a,b):
    if b==0:
        return abs(a)
    return gcd(b,a%b)

gcd(48,18)
```

Binary Search Using Recursion

 

```python
def bin(arr,low,high,target):
    if low>high:
        return -1
    mid=(low+high)//2
    if arr[mid]==target:
        return mid

    elif target<arr[mid]:
        return bin(arr,low,mid-1,target)

    else:
        return bin(arr,mid+1,high,target)
```

Program 23. Tower of Hanoi for n Disks
 
  

```python
def TOH(n,start,aux,end):
    if n==1:
        print("Move disk 1 from {} to {}".format(start,end))
        return 
    TOH(n-1,start,end,aux)
    print("Move disk {} from {} to {}".format(n,start,end))
    TOH(n-1,aux,start,end)
TOH(2,"A","B","C")
```

Program 24. Reverse a Number Using Recursion

 
 Aim: To reverse the digits of an integer using recursion.

```python
def reverse(n,result=0):
    if n==0:
        return result
    return reverse(n//10,result*10+n%10)
reverse(103010)
```

Program 25. Sum of Digits of a Number

```python
def sum(n):
    if n==0:
        return 0
    return n%10+sum(n//10)
sum(1024)
```

Program 27. Sum of Array Elements Using Recursion

  

```python
def array_sum(arr,index):
    if index==len(arr):
        return 0
    return arr[index]+array_sum(arr,index+1)
array_sum([1,2,3,4],0)
```
